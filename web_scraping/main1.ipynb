```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://www.bbc.com/sport/cricket/mens-england-division-one-multi-day/table"

response = requests.get(url)
```

```python
response.status_code
```

```python
soup = BeautifulSoup(response.content, 'html.parser')
```

```python
print(soup.prettify())  # This will print the HTML content in a readable format
```

```python
Teams = []
played = []
won = []
lost = []
```

```python
all_teams = soup.find('tbody')
print(all_teams.prettify())
```

```python
for each_team in all_teams:
    team_name = each_team.find('div', class_='ssrcss-1m4x485-DesktopWidthOnly e1sk6csp2').get_text(strip=True)
    total_played = each_team.find('td',class_ ='ssrcss-126tggj-StyledTableData ef9ipf1').get_text(strip=True)
    total_wins = each_team.find('td',class_='ssrcss-1jyl0o1-StyledTableData ef9ipf1').get_text(strip=True)
    # total_loose = each_team.find('td',class_ = 'ssrcss-1jyl0o1-StyledTableData ef9ipf1')
    losses = each_team.find_all('td')
    if len(losses) > 6:
        total_loose =int(losses[6].get_text(strip=True))
        print(total_loose)
        # break

    
    Teams.append(team_name)
    played.append(total_played)
    won.append(total_wins)
    lost.append(total_loose)
    
```

```python
data = {
        'Teams': Teams,
        'Played': played,
        'Won': won,
        'Lost': lost
}

df_teams = pd.DataFrame(data)
```
